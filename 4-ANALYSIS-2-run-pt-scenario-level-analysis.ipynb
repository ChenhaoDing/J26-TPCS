```python
import pandas as pd
import numpy as np
import os
from tqdm import tqdm
from itertools import product

from src.analysis.pt import process_pt_scenario_level_stats
```

# Variables

```python
# Demand Size
demand_size_list = [i for i in range(100, 1001, 100)]
# Demand Split Ratio (Intra Modal, %)
demand_split_ratio_list = [0, 20, 40, 60, 80]

# Random Seed
random_seed_list = [3, 6, 9]

# Network Name
network_name = "11-500"

# PT Network Name list
pt_network_name_list = ["basic", "ring", "diagonal", "mature"]

# Train headway (minutes)
train_headway_list = [10, 20, 30]

# Bus headway (minutes)
bus_headway_list = [10]

# Demand File Directory
demand_dir = f"data/demand/{network_name}/pt/"

# Results Output Directory
output_dir = f"data/pt-sim-results/"
```

# Scenario level

```python
all_demand_combinations = list(product(demand_size_list, demand_split_ratio_list, random_seed_list))

all_results = []

for pt_network_name in pt_network_name_list:
    for train_headway in train_headway_list:
        for bus_headway in bus_headway_list:
            scenario_name = f"{pt_network_name}_pt_train-{train_headway}_bus-{bus_headway}"
            scenario_results_dir = os.path.join(output_dir, scenario_name)
            print(f"Running PT scenario level analysis for scenario: {scenario_name}")

            for demand_size, demand_split_ratio, random_seed in tqdm(all_demand_combinations):
                demand_name = f"pt_ds{demand_size}_dsr{demand_split_ratio}_rs{random_seed}.csv"
                results_df = pd.read_csv(os.path.join(scenario_results_dir, demand_name))

                scenario_parameters = {
                    "pt_network_name": pt_network_name,
                    "train_headway": train_headway,
                    "bus_headway": bus_headway,
                    "demand_size": demand_size,
                    "intra_ratio": demand_split_ratio,
                    "random_seed": random_seed
                }

                result_dict = process_pt_scenario_level_stats(results_df, scenario_parameters)
                all_results.append(result_dict)

# Convert all results to DataFrame
results_summary_df = pd.DataFrame(all_results)

# Save the summary results to CSV
summary_output_path = os.path.join(output_dir, "all_pt_scenario_level_analysis_results.csv")
results_summary_df.to_csv(summary_output_path, index=False)
print(f"Saved PT scenario level analysis results to {summary_output_path}")
```
